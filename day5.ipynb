```python
import pandas as pd
import numpy as np

students = pd.DataFrame({
    "student_id": [1, 2, 3, 4, 5],
    "name": ["Maria", "James", "Priya", "Tomas", "Lena"]
})

scores = pd.DataFrame({
    "student_id": [1, 2, 3, 4, 6],
    "score": [92, 78, 85, 63, 99]
})

merged = pd.merge(students, scores, on="student_id")
print(merged)
```

```python
left_merged = pd.merge(students, scores, on="student_id", how="left")
print(left_merged)

outer_merged = pd.merge(students, scores, on="student_id", how="outer")
print(outer_merged)
```

```python
print(outer_merged.isna())
print(outer_merged.isna().sum())

filled = outer_merged.fillna({"score": 0, "name": "Unknown"})
print(filled)

dropped = outer_merged.dropna()
print(dropped)
```

```python
#Create a new DataFrame called attendance with columns student_id and days_present, use values [1, 2, 3, 4] for student_id and any four numbers you want for days_present
attendance = pd.DataFrame({
    "student_id": [1, 2, 3, 4],
    "days_present": [6,  9, 7, 8]
})
print(attendance)

#Merge students, scores, and attendance together into a single DataFrame using an outer join, so no student from any of the three tables gets dropped, hint, you'll need two separate merge calls chained together, merging two tables first, then merging that result with the third
all_merged= pd.merge(pd.merge(students, scores, on="student_id", how="outer"), attendance, on="student_id", how="outer")
print(all_merged)
#Fill any missing days_present values with 0, but leave missing score and name values as NaN
print(all_merged.fillna({"days_present": 0}))
```

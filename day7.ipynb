```python
import numpy as np
import pandas as pd
from scipy import stats

scores = np.array([72, 85, 90, 63, 88, 55, 91, 76, 68, 95])

print(np.mean(scores))
print(np.median(scores))
print(np.std(scores, ddof=1))
print(np.var(scores, ddof=1))
print(stats.describe(scores))
```

```python
n = len(scores)
mean = np.mean(scores)
std_err = stats.sem(scores)

confidence_level = 0.95
degrees_freedom = n - 1

confidence_interval = stats.t.interval(confidence_level, degrees_freedom, loc=mean, scale=std_err)
print(f"Mean: {mean}")
print(f"Standard Error: {std_err}")
print(np.std(scores, ddof=1) / np.sqrt(n))
print(f"95% Confidence Interval: {confidence_interval}")
```

```python
hypothesized_mean = 75

t_statistic, p_value = stats.ttest_1samp(scores, hypothesized_mean)
print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value}")

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
#Problem
#A campus coffee shop claims their average wait time is 4 minutes. You suspect it's actually longer. You time 12 customers and record these wait times, in minutes:
wait_times = np.array([4.5, 5.1, 3.8, 4.9, 5.3, 4.2, 4.7, 5.5, 4.1, 4.8, 5.0, 4.6])

# H0: mu = 4, H1: mu > 4
#Compute the sample mean and a 95% confidence interval for the true mean wait time
mean = np.mean(wait_times)
print(mean)
ci = stats.t.interval(.95, len(wait_times) -1, loc= mean, scale = stats.sem(wait_times))
print(f"95% Confidence Interval: {ci}")

#Run a one-sample hypothesis test against the claimed value of 4 minutes, make a conclusion
print(stats.ttest_1samp(wait_times, 4, alternative="greater"))
```

```python
# Conclude: Reject the null hypothesis, the population wait time is greater than 4 minutes.
```

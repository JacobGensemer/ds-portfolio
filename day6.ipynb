```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Maria", "James", "Priya", "Tomas", "Lena"],
    "score": [92, 78, 85, 63, 95],
    "hours_studied": [8, 4, 6, 2, 9]
})

plt.bar(df["name"], df["score"])
plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Scores by Student")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["hours_studied"], df["score"])
plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.title("Score vs Hours Studied")
plt.show()
```

```python
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="hours_studied", y="score", hue="name", s=150)
plt.title("Score vs Hours Studied")
plt.show()
```

```python
#Create a seaborn bar plot showing score by name.
sns.barplot(data = df, x = "name", y = "score")

#Add a horizontal reference line at score = 70 across the chart, marking the passing threshold visually.
plt.axhline(y=70)

#Give the chart a title and axis labels
plt.title("Score by Student")
plt.xlabel("Name")
plt.ylabel("Score")
plt.show()
```

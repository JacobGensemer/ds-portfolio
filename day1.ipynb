```python
scores = [72, 85, 90, 63, 88, 55]
for score in scores:
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    else:
        grade = "F"

    print(f"score: {score}, Grade: {grade}")
```

```python
def classify_score(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >=70:
        return "C"
    else:
        return "F"

for score in scores:

    print(f"Score: {score}, Grade: {classify_score(score)}")
```

```python
def is_passing(score):
    if score >= 70:
        return True
    else:
        return False

for score in scores:
 if is_passing(score):
    print(score)
```

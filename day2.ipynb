```python
students = ["Maria", "James", "Priya", "Tomas"]
students.append("Lena")
print(students)
students.remove("James")
print(students)
print(students[0])
print(students[-1])
print(students[1:3])
```

```python
grades = {"Maria": 92, "Priya": 85, "Tomas": 78}
grades["Lena"] = 95
print(grades)
print(grades["Priya"])

for name, score in grades.items():
    print(f"{name}: {score}")

print("James" in grades)
print("Maria" in grades)
```

```python
scores = [72, 85, 90, 63, 88, 55]
squared = [s ** 2 for s in scores]
print(squared)

passing = [s for s in scores if s >= 70]
print(passing)

grade_map = {s: ("Pass" if s >= 70 else "Fail") for s in scores}
print(grade_map)
```

```python
who_passed = [a for a in students if (a in grades and grades[a] >= 70)]


print(who_passed)
```

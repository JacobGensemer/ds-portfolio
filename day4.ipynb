```python
import pandas as pd

data = {
    "name": ["Maria", "James", "Priya", "Tomas", "Lena"],
    "score": [92, 78, 85, 63, 95],
    "hours_studied": [8, 4, 6, 2, 9]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
print(df.shape)
print(df.dtypes)
```

```python
print(df["name"])
print(df[["name", "score"]])
print(df.iloc[0])
print(df.loc[2])
print(df[df["score"] >= 80])
```

```python
df["passed"] = df["score"] >= 70
print(df)

df["score_per_hour"] = df["score"] / df["hours_studied"]
print(df)

print(df["score"].mean())
print(df.groupby("passed")["score"].mean())
```

```python
#Create a new column called efficient that's True if a student's score_per_hour is above 15, False otherwise
df["efficient"] = df["score_per_hour"] > 15
print(df)

#Sort the DataFrame by score in descending order, highest score first, and print the result
print(df.sort_values(by =["score"], ascending = False ))

#Using .loc, select and print only the name and score columns for students where hours_studied is less than 5
print(df.loc[df["hours_studied"] < 5,["name", "score"]])
```
